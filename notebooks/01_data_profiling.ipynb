### Finding: Campaign Attribution Risk

13,235 of 23,344 targeted accounts were associated with multiple
campaigns (~56.7%).

Therefore, attributing a payment to the latest campaign alone could
misrepresent campaign performance.

A time-based attribution window is required.

```python
payments["event_at"] = pd.to_datetime(payments["event_at"])
targeting["target_date"] = pd.to_datetime(targeting["target_date"])

payment_target = payments.merge(
    targeting[["account_id", "campaign_id", "target_date"]],
    on="account_id",
    how="left"
)

payment_target["days_after_target"] = (
    payment_target["event_at"].dt.normalize()
    - payment_target["target_date"]
).dt.days

print(payment_target["days_after_target"].describe())
```

A negative value means the payment happened before the targeting event.

But there's an important issue: our merge matched every payment with every targeting record for that account, so this distribution includes many irrelevant campaign-targeting events.

```python
targeting = targeting.copy()
payments = payments.copy()

targeting["target_date"] = pd.to_datetime(targeting["target_date"])
payments["event_at"] = pd.to_datetime(payments["event_at"])

# Remove missing keys just in case
targeting = targeting.dropna(subset=["account_id", "target_date"])
payments = payments.dropna(subset=["account_id", "event_at"])

# Sort primarily by the time column
targeting = targeting.sort_values(["target_date", "account_id"])
payments = payments.sort_values(["event_at", "account_id"])

payment_target = pd.merge_asof(
    payments,
    targeting[["account_id", "campaign_id", "target_date"]],
    left_on="event_at",
    right_on="target_date",
    by="account_id",
    direction="backward"
)

payment_target["days_after_target"] = (
    payment_target["event_at"].dt.normalize()
    - payment_target["target_date"]
).dt.days

print(payment_target["days_after_target"].describe())
```

```python
for days in [7, 14, 30, 60, 90]:
    count = (payment_target["days_after_target"] <= days).sum()
    print(f"{days} days: {count} payments")
```

```python
print(payment_target["days_after_target"].notna().sum())
```

```python
for days in [7, 14, 30, 60, 90]:
    count = (
        payment_target["days_after_target"].notna()
        & (payment_target["days_after_target"] <= days)
    ).sum()
    print(f"{days} days: {count} payments")
```

```python
borrower_conflict_ids = (
    borrowers_clean.groupby("borrower_id")
    .agg(
        name_n=("name", "nunique"),
        phone_n=("phone", "nunique"),
        email_n=("email", "nunique"),
        city_n=("city", "nunique"),
        state_n=("state", "nunique")
    )
)

borrower_conflict_ids["identity_conflict"] = (
    borrower_conflict_ids[
        ["name_n", "phone_n", "email_n", "city_n", "state_n"]
    ].max(axis=1) > 1
)

print(
    "Borrower IDs with identity conflicts:",
    borrower_conflict_ids["identity_conflict"].sum()
)
```

## 15 . agents table 

```python
agents = datasets["agents"].copy()

print("Original rows:", len(agents))
print("Exact duplicate rows:", agents.duplicated().sum())
print(
    "Duplicate agent IDs:",
    agents["agent_id"].duplicated().sum()
)
```

```python
agent_counts = agents["agent_id"].value_counts()

print(agent_counts.describe())
print()
print(agent_counts.value_counts().sort_index())
```

```python
sample_agent = agent_counts.index[0]

print(
    agents[agents["agent_id"] == sample_agent]
)
```

```python
agent_identity = (
    agents.groupby("agent_id")
    .agg(
        employee_codes=("employee_code", "nunique"),
        names=("agent_name", "nunique"),
        vendors=("vendor_id", "nunique"),
        teams=("team", "nunique"),
        statuses=("status", "nunique")
    )
)

print("Agents with multiple employee codes:",
      (agent_identity["employee_codes"] > 1).sum())

print("Agents with multiple names:",
      (agent_identity["names"] > 1).sum())

print("Agents with multiple vendors:",
      (agent_identity["vendors"] > 1).sum())

print("Agents with multiple teams:",
      (agent_identity["teams"] > 1).sum())

print("Agents with multiple statuses:",
      (agent_identity["statuses"] > 1).sum())
```

```python
agents["joined_at"] = pd.to_datetime(
    agents["joined_at"],
    errors="coerce"
)

agents["updated_at"] = pd.to_datetime(
    agents["updated_at"],
    errors="coerce"
)

print(
    "Invalid joined_at:",
    agents["joined_at"].isna().sum()
)

print(
    "Invalid updated_at:",
    agents["updated_at"].isna().sum()
)

print(
    "joined_at > updated_at:",
    (agents["joined_at"] > agents["updated_at"]).sum()
)
```

```python
agent_conflict_ids = agent_identity[
    (agent_identity["employee_codes"] > 1) |
    (agent_identity["names"] > 1) |
    (agent_identity["vendors"] > 1) |
    (agent_identity["teams"] > 1) |
    (agent_identity["statuses"] > 1)
].index

agents_clean = agents.copy()

agents_clean["identity_conflict"] = (
    agents_clean["agent_id"].isin(agent_conflict_ids)
)

agents_clean["joined_after_updated"] = (
    agents_clean["joined_at"] >
    agents_clean["updated_at"]
)

print(agents_clean["identity_conflict"].value_counts())
print()
print(agents_clean["joined_after_updated"].value_counts())
```

```python
agents_clean.to_csv(
    "../data/processed/agents_clean.csv",
    index=False
)

print("Saved agents_clean.csv")
```

```python
import os

print(
    "File exists:",
    os.path.exists("../data/processed/agents_clean.csv")
)
```
